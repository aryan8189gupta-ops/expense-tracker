```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
dataset = pd.read_csv("BostonHousing.csv")

x = dataset[["indus"]].head(5)
y = dataset[["nox"]].head(5) 


ss = StandardScaler()
mmmm = ss.fit_transform(x)

mo = ss.fit_transform(y)
m = pd.DataFrame(mmmm, columns=["indus"])
k = pd.DataFrame(mo,columns=["nox"])
result = pd.concat([m,k],axis=1)
print(result)
 
```
